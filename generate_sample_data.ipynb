```python
import open3d as o3d
```

```python
import numpy as np

# 10x10x10 박스의 모서리 좌표 생성
size = 5
corners = np.array([
    [0, 0, 0],
    [0, 0, size],
    [0, size, 0],
    [0, size, size],
    [size, 0, 0],
    [size, 0, size],
    [size, size, 0],
    [size, size, size]
])

# 각 모서리를 따라 점 생성 (간격: 1)
points = []
for i in range(8):
    for j in range(i+1, 8):
        diff = np.abs(corners[i] - corners[j])
        if np.count_nonzero(diff) == 1:  # 한 축만 다르면 모서리
            axis = np.where(diff != 0)[0][0]
            for k in range(11):
                pt = corners[i].copy()
                pt[axis] += (corners[j][axis] - corners[i][axis]) * k / 10
                points.append(pt)

points = np.unique(np.array(points), axis=0)

# open3d 포인트 클라우드로 변환
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)
o3d.io.write_point_cloud("data/sample/box_edges.pcd", pcd, write_ascii=True)
```

```python
import plotly.graph_objs as go

fig = go.Figure(data=[go.Scatter3d(
    x=points[:, 0],
    y=points[:, 1],
    z=points[:, 2],
    mode='markers',
    marker=dict(size=3)
)])
fig.write_html("data/sample/box_edges_plot.html")
```
